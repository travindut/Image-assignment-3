```python
import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt

print(f"PyTorch version: {torch.__version__}")

# 1. LeNet-5 Architecture Definition

class LeNet5(nn.Module):
    def __init__(self, num_classes=10):
        super(LeNet5, self).__init__()
        # Input shape: (batch_size, 1, 32, 32)
        self.feature_extractor = nn.Sequential(
            # C1: (1, 32, 32) -> (6, 28, 28)
            nn.Conv2d(in_channels=1, out_channels=6, kernel_size=5, stride=1),
            nn.Tanh(),  # Classic LeNet used Tanh
            # S2: (6, 28, 28) -> (6, 14, 14)
            nn.MaxPool2d(kernel_size=2, stride=2),
            # C3: (6, 14, 14) -> (16, 10, 10)
            nn.Conv2d(in_channels=6, out_channels=16, kernel_size=5, stride=1),
            nn.Tanh(),
            # S4: (16, 10, 10) -> (16, 5, 5)
            nn.MaxPool2d(kernel_size=2, stride=2),
        )

        self.classifier = nn.Sequential(
            # C5: Flatten 16 * 5 * 5 = 400 -> 120
            nn.Linear(in_features=16 * 5 * 5, out_features=120),
            nn.Tanh(),
            # F6: 120 -> 84
            nn.Linear(in_features=120, out_features=84),
            nn.Tanh(),
            # Output layer: 84 -> num_classes
            nn.Linear(in_features=84, out_features=num_classes),
        )

    def forward(self, x):
        x = self.feature_extractor(x)
        x = torch.flatten(x, 1)  # Flatten all dimensions except batch
        logits = self.classifier(x)
        return logits


# 2. Data Preparation & Preprocessing

# MNIST is 28x28; Pad(2) makes it 32x32 to match LeNet-5 input dimensions
transform = transforms.Compose([
    transforms.Pad(padding=2),
    transforms.ToTensor(),
])

batch_size = 50
data_root = "./data"

trainset = torchvision.datasets.MNIST(
    root=data_root, train=True, download=True, transform=transform
)
testset = torchvision.datasets.MNIST(
    root=data_root, train=False, download=True, transform=transform
)

trainloader = torch.utils.data.DataLoader(
    trainset, batch_size=batch_size, shuffle=True
)
testloader = torch.utils.data.DataLoader(
    testset, batch_size=batch_size, shuffle=False
)


# 3. Model, Loss, Optimizer Initialization


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print(f"Using device: {device}")

model = LeNet5().to(device)
print(model)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

batch_losses = []
train_losses = []
train_accuracies = []
test_accuracies = []


def accuracy(loader):
    model.eval()  # Set model to evaluation mode
    correct, total = 0, 0
    with torch.no_grad():
        for data, labels in loader:
            data, labels = data.to(device), labels.to(device)
            outputs = model(data)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    return 100 * correct / total


# 4. Training Loop

epochs = 10

for epoch in range(epochs):
    model.train()  # Set model back to train mode
    running_loss = 0.0

    for i, (images, labels) in enumerate(trainloader):
        images, labels = images.to(device), labels.to(device)

        # Forward pass
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)

        # Backward pass & optimization
        loss.backward()
        optimizer.step()

        # Track losses
        running_loss += loss.item()
        batch_losses.append(loss.item())

    avg_loss = running_loss / len(trainloader)
    train_losses.append(avg_loss)

    # Evaluate epoch performance
    train_acc = accuracy(trainloader)
    test_acc = accuracy(testloader)
    train_accuracies.append(train_acc)
    test_accuracies.append(test_acc)

    print(
        f"Epoch {epoch + 1}/{epochs} | "
        f"Loss: {avg_loss:.4f} | "
        f"Train Acc: {train_acc:.2f}% | "
        f"Test Acc: {test_acc:.2f}%"
    )

# ---------------------------------------------------------
# 5. Visualizations
# ---------------------------------------------------------
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(batch_losses, label="Batch Loss")
plt.title("Batch Loss")
plt.xlabel("Batch")
plt.ylabel("Loss")
plt.legend()

plt.subplot(1, 3, 2)
plt.plot(range(1, epochs + 1), train_losses, label="Train Loss", color="orange")
plt.title("Train Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.subplot(1, 3, 3)
plt.plot(
    range(1, epochs + 1),
    train_accuracies,
    label="Train Accuracy",
    color="green",
)
plt.plot(
    range(1, epochs + 1), test_accuracies, label="Test Accuracy", color="red"
)
plt.title("Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

num_classes = 10
cm = torch.zeros(num_classes, num_classes, dtype=torch.int64)

# 1. Accumulate predictions directly into a PyTorch tensor
model.eval()
with torch.no_grad():
    for images, labels in testloader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, preds = torch.max(outputs, 1)

        # Accumulate counts at [ground_truth, prediction] coordinates
        for t, p in zip(labels.view(-1), preds.view(-1)):
            cm[t.long(), p.long()] += 1

cm_np = cm.cpu().numpy()

# 2. Plot heatmap
fig, ax = plt.subplots(figsize=(8, 8))
cax = ax.imshow(cm_np, interpolation="nearest", cmap=plt.cm.Blues)
plt.title("Confusion Matrix (Test Set)", fontsize=14)
plt.colorbar(cax, fraction=0.046, pad=0.04)

tick_marks = range(num_classes)
ax.set_xticks(tick_marks)
ax.set_yticks(tick_marks)
ax.set_xticklabels(tick_marks)
ax.set_yticklabels(tick_marks)

# 3. Annotate cell values with contrasting text color
thresh = cm_np.max() / 2.0
for i in range(num_classes):
    for j in range(num_classes):
        val = cm_np[i, j]
        color = "white" if val > thresh else "black"
        ax.text(j, i, f"{val}", ha="center", va="center", color=color, fontsize=9)

ax.set_ylabel("True Label", fontsize=12)
ax.set_xlabel("Predicted Label", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import torchvision.transforms as transforms

print(f"PyTorch version: {torch.__version__}")

# ---------------------------------------------------------
# 1. Modern 3-Layer CNN with BatchNorm & Dropout
# ---------------------------------------------------------
class ModernCNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        # --- Layer 1: (1, 32, 32) -> (16, 16, 16) ---
        self.conv1 = nn.Conv2d(in_channels=1, out_channels=16, kernel_size=3, padding=1)
        self.bn1 = nn.BatchNorm2d(16)

        # --- Layer 2: (16, 16, 16) -> (32, 8, 8) ---
        self.conv2 = nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm2d(32)

        # --- Layer 3: (32, 8, 8) -> (64, 4, 4) ---
        self.conv3 = nn.Conv2d(in_channels=32, out_channels=64, kernel_size=3, padding=1)
        self.bn3 = nn.BatchNorm2d(64)

        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)

        # --- Fully Connected Classifier ---
        # 64 channels * 4 height * 4 width = 1024
        self.fc1 = nn.Linear(64 * 4 * 4, 128)
        self.bn4 = nn.BatchNorm1d(128)
        self.dropout1 = nn.Dropout(p=0.5)

        self.fc2 = nn.Linear(128, 64)
        self.dropout2 = nn.Dropout(p=0.5)

        self.fc3 = nn.Linear(64, num_classes)

    def forward(self, x):
        # Conv Block 1
        x = self.pool(F.relu(self.bn1(self.conv1(x))))

        # Conv Block 2
        x = self.pool(F.relu(self.bn2(self.conv2(x))))

        # Conv Block 3
        x = self.pool(F.relu(self.bn3(self.conv3(x))))

        # Flatten features: (batch_size, 1024)
        x = torch.flatten(x, 1)

        # Classifier
        x = self.dropout1(F.relu(self.bn4(self.fc1(x))))
        x = self.dropout2(F.relu(self.fc2(x)))
        x = self.fc3(x)
        return x


# 2. Data Preparation & Preprocessing
transform = transforms.Compose([
    transforms.Pad(padding=2),
    transforms.ToTensor(),
])

batch_size = 50
data_root = "./data"

trainset = torchvision.datasets.MNIST(
    root=data_root, train=True, download=True, transform=transform
)
testset = torchvision.datasets.MNIST(
    root=data_root, train=False, download=True, transform=transform
)

trainloader = torch.utils.data.DataLoader(
    trainset, batch_size=batch_size, shuffle=True
)
testloader = torch.utils.data.DataLoader(
    testset, batch_size=batch_size, shuffle=False
)

# 3. Model, Loss, Optimizer Initialization
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print(f"Using device: {device}")

model = ModernCNN().to(device)
print(model)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

batch_losses = []
train_losses = []
train_accuracies = []
test_accuracies = []


def accuracy(loader):
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for data, labels in loader:
            data, labels = data.to(device), labels.to(device)
            outputs = model(data)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    return 100 * correct / total


# 4. Training Loop
epochs = 10

for epoch in range(epochs):
    model.train()
    running_loss = 0.0

    for i, data in enumerate(trainloader):
        inputs, labels = data
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        batch_losses.append(loss.item())

    avg_loss = running_loss / len(trainloader)
    train_losses.append(avg_loss)

    train_acc = accuracy(trainloader)
    test_acc = accuracy(testloader)
    train_accuracies.append(train_acc)
    test_accuracies.append(test_acc)

    print(
        f"Epoch {epoch + 1}/{epochs} | "
        f"Loss: {avg_loss:.4f} | "
        f"Train Acc: {train_acc:.2f}% | "
        f"Test Acc: {test_acc:.2f}%"
    )

# 5. Visualizations

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(batch_losses, label="Batch Loss")
plt.title("Batch Loss")
plt.xlabel("Batch")
plt.ylabel("Loss")
plt.legend()

plt.subplot(1, 3, 2)
plt.plot(range(1, epochs + 1), train_losses, label="Train Loss", color="orange")
plt.title("Train Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.subplot(1, 3, 3)
plt.plot(range(1, epochs + 1), train_accuracies, label="Train Accuracy", color="green")
plt.plot(range(1, epochs + 1), test_accuracies, label="Test Accuracy", color="red")
plt.title("Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()

plt.tight_layout()
plt.show()
```
